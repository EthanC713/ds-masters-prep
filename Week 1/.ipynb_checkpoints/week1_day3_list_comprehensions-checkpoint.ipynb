```python
# The long way - using a for loop

squares = []
for n in range(1,6):
    squares.append(n**2)
print(squares) # [1,4,9,16,25]
```

```python
# The short way - list comprehensions

squares = [n**2 for n in range(1,6)]
print(squares) # [1,4,9,16,25]

# the pattern is always:
# [expression for item in iterable]
```

```python
# Practice exercise:
# Rewrite the following loops as list comprehensions:

#1. A list of numbers 1-10 multiplied by 3

multiples = []
for n in range(1, 11):
    multiples.append(n * 3)
print(multiples) 

multiples = [n*3 for n in range(1,11)]
print(multiples)
```

```python
#2. A list of the lengths of each word

words = ["python", "data", "science", "machine", "learning"]
lengths = []
for word in words:
    lengths.append(len(word))
print(lengths)

lengths = [len(word) for word in words]
print(lengths)
```

```python
#3. A list of all numbers 1-20 squared

squared = []
for n in range(1, 21):
    squared.append(n**2)
print(squared)

squared = [n**2 for n in range(1,21)]
print(squared)
```

```python
# Adding a condition - the pattern is:
# [expression for item in iterable if condition]

# Only even numbers from 1-20
evens = [n for n in range(1,21) if n%2 == 0]
print(evens)

# Only words longer than 4 characters
words = ["python", "data", "science", "AI", "ML"]
long_words = [word for word in words if len(word) > 4]
print(long_words)
```

```python
# Practice:
gpas = [3.8, 2.9, 3.5, 3.1, 4.0, 2.7, 3.9, 3.3, 2.5, 3.6]

# Using list comprehensions:
# 1. GPAs above 3.5
# 2. GPAs rounded to 1 decimal place
# 3. GPAs filtered to only failing GPAs (< 2.0)

above_threefive = [gpa for gpa in gpas if gpa > 3.5]
print(above_threefive)

one_decimal = [round(gpa,1) for gpa in gpas]
print(one_decimal)

failing_gpas = [gpa for gpa in gpas if gpa < 2.0]
print(failing_gpas)
```

```python
# List comprehensions with functions

def classify_grade(grade):
    if grade >= 90:
        return "A"
    elif grade >= 80:
        return "B"
    elif grade >= 70:
        return "C"
    else:
        return "F"

grades = [92, 88, 75, 100, 61, 79]

# Call classify_grade on every item in one line
classifications = [classify_grade(g) for g in grades]
print(classifications) 
```

```python
# Practice
students = ["ada", "alan", "grace", "turing"]
gpas =     [3.8,   3.1,    2.9,     4.0]

# Using list comprehensions and your functions from Day 2:
# 1. Capitalize every student name using .upper()
# 2. Classify each GPA — write a classify_gpa function first that returns
#    "High" (>= 3.5), "Mid" (>= 2.5), or "Low" (< 2.5)
# 3. Combine both — create a list of strings like "ADA: High"

#1:
upper_case = [name.upper() for name in students]
print(upper_case)

#2:
def classify_gpa(grade):
    if grade >= 3.5:
        return "High"
    elif grade >= 2.5:
        return "Mid"
    else:
        return "Low"

classified = [classify_gpa(g) for g in gpas]
print(classified)

#3:
combined = [f"{name.upper()}: {classify_gpa(grades)}" for name,grades in zip(students, gpas)]
print(combined)
```

```python
# Exercise:

def gradebook(name, grades):
    avg_grade = round(sum(grades) / len(grades), 2)
    if avg_grade >= 90:
        return f"{name}: average {avg_grade} → A"
    elif avg_grade >= 80:
        return f"{name}: average {avg_grade} → B"
    elif avg_grade >= 70:
        return f"{name}: average {avg_grade} → C"
    else:
        return f"{name}: average {avg_grade} → F"

students = [
    {"name": "Ada",    "grades": [92, 88, 85, 90]},
    {"name": "Alan",   "grades": [70, 65, 72, 68]},
    {"name": "Grace",  "grades": [55, 60, 58, 50]},
    {"name": "Turing", "grades": [95, 98, 100, 97]},
    {"name": "Von",    "grades": [40, 45, 38, 50]}
]

# Using list comprehensions, create:
# 1. A list of just the student names
# 2. A list of each student's average grade (rounded to 2 decimal places)
# 3. A list of names of students who are passing (average >= 60)
# 4. A list of summary strings like "Ada: 88.75 → A"
#    using the previous gradebook function

#1: 
student_names = [student["name"] for student in students]
print(student_names)

#2:
avg_grades = [round(sum(student["grades"])/len(student["grades"]),2) for student in students]
print(avg_grades)

#3: 
passing_students = [student["name"] for student in students if sum(student["grades"])/len(student["grades"]) >= 60]
print(passing_students)

#4:
summary = [gradebook(student["name"],student["grades"]) for student in students]
print(summary)
```
