```python
# Basic if / elif / else

grade = 88

if grade > 90:
    print ("A")
elif grade >= 80:
    print ("B")
elif grade >= 70:
    print ("C")
else:
    print ("Below C")
```

```python
# Combining conditions with and / or
age = 20
has_id = True

if age >= 18 and has_id:
    print ("Entry allowed")
else:
    print ("Entry denied")
```

```python
# loop over a range
for i in range(5):
    print (i) # prints 0,1,2,3,4

# loop over a list
fruits = ["apple","banana","cherry"]
for fruit in fruits:
    print (fruit)

# loop with index using enumerate
for index, fruit in enumerate(fruits):
    print(f"{index}:{fruit}")
```

```python
# while loop - runs as long as a condition is true
count = 0
while count < 5:
    print (count)
    count += 1
```

```python
# break out of loop early

count = 0
while True:
    print (count)
    count += 1
    if count == 5:
        break
```

```python
# FizzBuzz Exercise: 

for number in range(21):
    print (number)
    
    if number % 3 == 0:
        print ("Fizz")
    elif number % 5 == 0:
        print ("Buzz")
    elif number % 3 == 0 and number % 5 == 0:
        print ("FizzBuzz")
    
```

```python
# FizzBuzz Exercise: 

for number in range(1,21):
    print (number)
    
    if number % 3 == 0 and number % 5 == 0:
        print ("FizzBuzz")
    if number % 3 == 0:
        print ("Fizz")
    elif number % 5 == 0:
        print ("Buzz")
    else:
        print (number)
```

```python
for number in range(1, 21):
    if number % 3 == 0 and number % 5 == 0:
        print("FizzBuzz")
    elif number % 3 == 0:
        print("Fizz")
    elif number % 5 == 0:
        print("Buzz")
    else:
        print(number)
```

```python
# Practice
# Given this list of GPA scores:

gpas = [3.8, 2.9, 3.5, 3.1, 4.0, 2.7, 3.9]

# Using a for loop:
# 1. Print each GPA
# 2. Print only GPAs above 3.5
# 3. Count how many students have a GPA above 3.5

counter = 0

for gpa in gpas:    
    if gpa > 3.5:
        print(gpa)
        counter += 1
    else:
        print(gpa)

print(counter)
```

```python
gpas = [3.8, 2.9, 3.5, 3.1, 4.0, 2.7, 3.9]

# 1. Print each GPA
for gpa in gpas:
    print(gpa)

# 2 & 3. Print only GPAs above 3.5 and count them
counter = 0
for gpa in gpas:
    if gpa > 3.5:
        print(gpa)
        counter += 1

print(f"Number of students above 3.5 GPA: {counter}")
```
