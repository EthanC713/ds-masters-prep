```python
# A class is a blueprint for creating objects

class Student:
    # __init__ runs automatically when you create a new Student
    def __init__(self,name,gpa):
        self.name = name # Self stores the value on this specific object
        self.gpa = gpa

    def is_honors(self):
        if self.gpa >= 3.5:
            return f"{self.name} qualifies for honors"
        else:
            return f"{self.name} does not qualify"

# Creating objects from the class

s1 = Student("Ada",3.8)
s2 = Student("Alan",3.1)

# Accessing attributes

print(s1.name) #s1.name - an attribute is a value that is stored on the object
print(s1.gpa)

# Calling methods

print(s1.is_honors()) #s1.is_honors() - a method is a function that belongs to the object
print(s2.is_honors())
```

```python
# Practice:
# Build a class called Course with:
# - name, instructor, and credits as attributes
# - a method called describe() that returns a string like:
#   "Introduction to Python — taught by Ada, 3 credits"

class Course:
    def __init__(self,name,instructor,credits):
        self.name = name
        self.instructor = instructor
        self.credits = credits

    def describe(self):
        return f"{self.name} - taught by {self.instructor}, {self.credits} credits"

# Testing with these:
c1 = Course("Introduction to Python", "Ada", 3)
c2 = Course("Linear Algebra", "Turing", 4)
print(c1.describe())
print(c2.describe())
```

```python
# Example: Adding more methods

class Student:
    def __init__(self,name,gpa,courses=[]):
        self.name = name
        self.gpa = gpa
        self.courses = courses

    def is_honors(self):
        return self.gpa >= 3.5

    def enroll(self,course):
        self.courses.append(course)
        return f"{self.name} enrolled in {course}"

    def describe(self):
        if self.courses:
            course_list = ", ".join(self.courses)
            return f"{self.name} (GPA: {self.gpa}) - enrolled in {course_list}"
        else:
            return f"{self.name} (GPA: {self.gpa}) - no courses yet"

# Testing it:

s1 = Student("Ada", 3.8)
print(s1.describe())
print(s1.enroll("Introduction to Python"))
print(s1.enroll("Linear Algebra"))
print(s1.describe())
print(s1.is_honors())
```

```python
# Practice
# Add the following to your Course class:
# 1. A students attribute (empty list by default)
# 2. An enroll_student(name) method that adds a student to the list
#    and returns f"{name} enrolled in {course name}"
# 3. A roster() method that returns:
#    "Introduction to Python roster: Ada, Alan, Grace"
#    or "Introduction to Python — no students enrolled" if empty
# 4. A is_full() method that returns True if 30 or more students are enrolled

class Course:
    def __init__(self,name,instructor,credits,students=[]):
        self.name = name
        self.instructor = instructor
        self.credits = credits
        self.students = students

    def describe(self):
        return f"{self.name} - taught by {self.instructor}, {self.credits} credits"

    def enroll_student(self,name):
        self.students.append(name)
        return f"{name} enrolled in {self.name}"

    def roster(self):
        if self.students:
            students_list = ", ".join(self.students)
            return f"{self.name} roster: {students_list}"
        else:
            return f"{self.name} - no students enrolled"

    def is_full(self):
        return len(self.students) >= 30

# Testing it with:
c1 = Course("Introduction to Python", "Ada", 3)
print(c1.roster())
print(c1.enroll_student("Alan"))
print(c1.enroll_student("Grace"))
print(c1.enroll_student("Turing"))
print(c1.roster())
print(c1.is_full())
```

```python
# Using "None" instead of a mutable default argument students = []

class Course:
    def __init__(self,name,instructor,credits,students=None):  # Replacing the empty list default argument with None
        self.name = name
        self.instructor = instructor
        self.credits = credits
        self.students = students if students is not None else []

    def describe(self):
        return f"{self.name} - taught by {self.instructor}, {self.credits} credits"

    def enroll_student(self,name):
        self.students.append(name)
        return f"{name} enrolled in {self.name}"

    def roster(self):
        if self.students:
            students_list = ", ".join(self.students)
            return f"{self.name} roster: {students_list}"
        else:
            return f"{self.name} - no students enrolled"

    def is_full(self):
        return len(self.students) >= 30
```

```python
# Practice: 

# Build a Gradebook class that manages a collection of students
# Requirements:
# 1. __init__ sets up an empty students dictionary {}
# 2. add_student(name, grades) adds a student and their grades
# 3. average(name) returns the average grade for a student
# 4. top_students() returns a list of names with average >= 90
# 5. class_summary() prints a summary for every student like:
#    "Ada: average 88.75 → B"
#    "Turing: average 97.5 → A"

class Gradebook:
    def __init__(self,students=None):
        self.students = students if students is not None else {}

    def add_student(self,name,grades):
        self.students[name] = grades

    def average(self,name):
        students_avggrade = round(sum(self.students[name]) / len(self.students[name]),2)
        return students_avggrade

    def top_students(self):
        top = []
        for name,grades in self.students.items():
            if sum(grades) / len(grades) >= 90:
                top.append(name)
        return top

    def classify_grade(self,name,grades):
        avg_grade = round(sum(grades) / len(grades), 2)
        if avg_grade >= 90:
            return f"{name}: average {avg_grade} → A"
        elif avg_grade >= 80:
            return f"{name}: average {avg_grade} → B"
        elif avg_grade >= 70:
            return f"{name}: average {avg_grade} → C"
        else:
            return f"{name}: average {avg_grade} → F"

    def class_summary(self):
        for name,grades in self.students.items():
            print(self.classify_grade(name,grades))
            
# Testing it with this:
gb = Gradebook()
gb.add_student("Ada",    [92, 88, 85, 90])
gb.add_student("Alan",   [70, 65, 72, 68])
gb.add_student("Grace",  [55, 60, 58, 50])
gb.add_student("Turing", [95, 98, 100, 97])
gb.add_student("Von",    [40, 45, 38, 50])

print(gb.average("Ada"))
print(gb.top_students())
gb.class_summary()
```
