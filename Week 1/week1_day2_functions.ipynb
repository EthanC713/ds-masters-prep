```python
# Function with a default argument

def greet(name,greeting="Hello"):
    return f"{greeting}, {name}!"

print(greet("Ada"))  # uses default greeting
print(greet("Alan","Hi"))
```

```python
# Write a function called describe_student that:
# - Takes a name (string) and a gpa (float) as parameters
# - Has a default value of "Unknown" for name
# - Returns an f-string like: "Ada has a GPA of 3.8"

def describe_student(gpa,name = "Unknown"):
    return f"{name} has a GPA of {gpa}"

print(describe_student(3.8,"Ada"))
print(describe_student(3.5))
print(describe_student(0))
```

```python
def is_passing(grade):
    return grade >= 60    #returns True or False

def classify_grade(grade):
    if grade >= 90:
        return "A"
    elif grade >= 80:
        return "B"
    elif grade >= 70:
        return "C"
    else:
        return "F"

print(is_passing(75))        # True
print(is_passing(55))        # False
print(classify_grade(85))    # B
print(classify_grade(55))    # F
```

```python
# Practice

# Write a function called summarize that:
# - Takes a list of numbers
# - Returns a dictionary with keys "min", "max", and "average"

def summarize(scores):
    min_scores = min(scores)
    max_scores = max(scores)
    avg_scores = sum(scores)/len(scores)
    scores_dict = {"min": min_scores,"max": max_scores,"average": avg_scores}
    return scores_dict

scores = [88, 92, 75, 100, 61, 79]
result = summarize(scores)
print(result)
    
```

```python
# Variables defined inside a function 

def my_function():
    x = 10   # x only exists inside this function
    return x

print(my_function())  # works fine
# print(x) causes an error - x doesn't exist outside of function

# Global vs local

name = "Ada"  # Global variable

def change_name():
    name = "Alan" # Creates a new LOCAL variable, doesn't change the global
    print(f"Inside function: {name}")

change_name()
print(f"Outside function: {name}") # still Ada
```

```python
# Practice: 
# Build a function called gradebook that:
# 1. Takes a student name and a list of grades
# 2. Calculates the average grade
# 3. Classifies it as "A", "B", "C", or "F"
# 4. Returns a summary string like:
#    "Ada: average 88.5 → B"

def gradebook(name,grades):
    avg_grade = round(sum(grades)/len(grades),2)
    if avg_grade >= 90:
        return f"{name}: average {avg_grade} → A"
    elif avg_grade >= 80:
        return f"{name}: average {avg_grade} → B"
    elif avg_grade >= 70:
        return f"{name}: average {avg_grade} → C"
    elif avg_grade >= 60:
        return f"{name}: average {avg_grade} → D"
    else:
        return f"{name}: average {avg_grade} → F"

# Test it with these:
print(gradebook("Ada", [92, 88, 85, 90]))
print(gradebook("Alan", [70, 65, 72, 68]))
print(gradebook("Grace", [55, 60, 58, 50]))
```
