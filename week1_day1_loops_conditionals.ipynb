```python
# loop over a range
for i in range(5):
    print (i) # prints 0,1,2,3,4

# loop over a list
fruits = ["apple","banana","cherry"]
for fruit in fruits:
    print (fruit)

# loop with index using enumerate
for index, fruit in enumerate(fruits):
    print(f"{index}:{fruit}")
```

```python
# while loop - runs as long as a condition is true
count = 0
while count < 5:
    print (count)
    count += 1
```

```python
# break out of loop early

count = 0
while True:
    print (count)
    count += 1
    if count == 5:
        break
```

```python
# FizzBuzz Exercise: 

for number in range(21):
    print (number)
    
    if number % 3 == 0:
        print ("Fizz")
    elif number % 5 == 0:
        print ("Buzz")
    elif number % 3 == 0 and number % 5 == 0:
        print ("FizzBuzz")
    
```

```python
# FizzBuzz Exercise: 

for number in range(1,21):
    print (number)
    
    if number % 3 == 0 and number % 5 == 0:
        print ("FizzBuzz")
    if number % 3 == 0:
        print ("Fizz")
    elif number % 5 == 0:
        print ("Buzz")
    else:
        print (number)
```
