```python
# Creating tuples
point = (3, 5)
student = ("Ada", 3.8, "Computer Science")

# Accessing — same as lists
print(point[0])       # 3
print(student[-1])    # Computer Science

# Unpacking - very useful
x,y = point
print(f"x={x}, y={y}")

name,gpa,major = student
print(f"{name} studies {major}")

# Tuples are immutable - this causes an error
# point[0] = 10   # TypeError!

# Common use - returning multiple values from a function

def min_max(numbers):
    return min(numbers), max(numbers) # returns a tuple

low,high = min_max([88, 92, 75, 100, 61])
print(f"Min: {low}, Max: {high}")
```

```python
# 1. Create a tuple called coordinates with values (40.7128, 74.0060)
#    representing latitude and longitude of New York City
#    Unpack it into lat and lon variables and print them

coordinates = (40.7128, 74.0060)
lat,lon = coordinates
print(f"Latitude: {lat}, Longitude: {lon}")

# 2. Write a function called stats() that takes a list of numbers
#    and returns a tuple of (min, max, mean, median)
#    Hint: import statistics and use statistics.median()
#    Test it with: stats([88, 92, 75, 100, 61, 79])

import statistics
def stats(numbers):
    return min(numbers), max(numbers), statistics.mean(numbers), statistics.median(numbers)
print(stats([88, 92, 75, 100, 61, 79]))

# 3. Create a list of tuples representing students:
#    [("Ada", 3.8), ("Alan", 3.1), ("Grace", 2.9)]
#    Sort the list by GPA highest to lowest
#    Hint: use sorted() with key=lambda x: x[1], reverse=True

students = [("Ada", 3.8), ("Alan", 3.1), ("Grace", 2.9)]
sorted_students = sorted(students, key=lambda x: x[1], reverse=True)
print(sorted_students)
```

```python
# Creating sets
numbers = {1, 2, 3, 4, 5}
fruits = {"apple", "banana", "cherry"}

# Duplicates are automatically removed
grades = {88, 92, 75, 88, 100, 92}
print(grades)   # {88, 92, 75, 100} — duplicates gone

# Adding and removing
fruits.add("date")
fruits.remove("banana")

# Membership testing — much faster than lists for large data
print("apple" in fruits)    # True
print("banana" in fruits)   # False

# Set operations — very useful
set_a = {1, 2, 3, 4, 5}
set_b = {4, 5, 6, 7, 8}

print(set_a & set_b)   # intersection — {4, 5}
print(set_a | set_b)   # union — {1, 2, 3, 4, 5, 6, 7, 8}
print(set_a - set_b)   # difference — {1, 2, 3}
```

```python
# Practice:
enrolled_python  = ["Ada", "Alan", "Grace", "Turing", "Ada", "Alan"]
enrolled_algebra = ["Ada", "Turing", "Von", "Grace"]

# 1. Convert both lists to sets to remove duplicates

python_set = set(enrolled_python)
print(python_set)
algebra_set = set(enrolled_algebra)
print(algebra_set)

# 2. Find students enrolled in BOTH courses (intersection)

print(python_set & algebra_set)

# 3. Find students enrolled in EITHER course (union)

print(python_set | algebra_set)

# 4. Find students in Python but NOT in Algebra (difference)

print(python_set - algebra_set)

# 5. How many unique students are enrolled across both courses?

print(len(python_set | algebra_set))
```
