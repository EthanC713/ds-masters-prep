```python
# Quick review on classes and objects

# Build a class called Library with:
# - a name attribute
# - a books attribute (empty list by default)
# - an add_book(title, author) method that adds a dictionary
#   like {"title": "Python 101", "author": "Ada"} to the list
# - a search(author) method that returns all books by that author
# - a summary() method that prints:
#   "Library Name: 3 books available"

class Library:
    def __init__(self,name,books=None):
        self.name = name
        self.books = books if books is not None else []

    def add_book(self,title,author):
        book = {"title": title, "author": author}
        self.books.append(book)

    def search(self,author):
        return [book for book in self.books if book["author"] == author]

    def summary(self):
        return f"{self.name}: {len(self.books)} books available"

# Test it with:
lib = Library("City Library")
lib.add_book("Python 101", "Ada")
lib.add_book("Linear Algebra", "Turing")
lib.add_book("Statistics 101", "Ada")

print(lib.summary())
print(lib.search("Ada"))
```

```python
# Common list methods

numbers = [3, 1, 4, 1, 5, 9, 2, 6]

numbers.append(7) # adds to end
numbers.insert(0,99) # inserts at index 0
numbers.remove(1) # remove first instance of 1
numbers.pop() # removes and returns last item
numbers.sort() # sorts in place
numbers.reverse() # reverses in place

print(numbers)

# Slicing

fruits = ["apple", "banana", "cherry", "date", "elderberry"]
print(fruits[1:3])    # ["banana", "cherry"]
print(fruits[:2])     # ["apple", "banana"]
print(fruits[2:])     # ["cherry", "date", "elderberry"]
print(fruits[-1])     # "elderberry" — last item
print(fruits[::2])    # every other item
```

```python
grades = [88, 92, 75, 100, 61, 79, 95, 83, 70, 55]

# Using list methods:
# 1. Sort the grades from highest to lowest
# 2. Remove the lowest grade
# 3. Add a new grade of 91
# 4. Return the top 3 grades using slicing
# 5. Calculate the average of the remaining grades

#1. 
grades.sort(reverse=True)
#2. 
grades.pop()
#3. 
grades.append(91)
#4. 
grades.sort(reverse=True)
print(f"Top 3 grades: {grades[:3]}")
#5. 
average = round(sum(grades) / len(grades),2)
print(f"Average: {average}")
```

```python
# Common dictionary methods

student = {
    "name": "Ada",
    "gpa": 3.8,
    "courses": ["Python", "Linear Algebra"]
}

# Accessing values safely
print(student.get("name"))           # Ada
print(student.get("age", "Unknown")) # Unknown — default if key missing

# Looping
for key, value in student.items():
    print(f"{key}: {value}")

# Adding and updating
student["year"] = 2             # adds new key
student["gpa"] = 3.9            # updates existing key

# Removing
student.pop("year")             # removes key

# Check if key exists
if "name" in student:
    print("Name found")

# Dictionary of dictionaries
students = {
    "Ada":    {"gpa": 3.8, "year": 1},
    "Alan":   {"gpa": 3.1, "year": 2},
    "Grace":  {"gpa": 2.9, "year": 1}
}

# Access nested values
print(students["Ada"]["gpa"])    # 3.8
```

```python
# Practice

gradebook = {
    "Ada":    [92, 88, 85, 90],
    "Alan":   [70, 65, 72, 68],
    "Grace":  [55, 60, 58, 50],
    "Turing": [95, 98, 100, 97]
}

# 1. Add "Von" with grades [40, 45, 38, 50]
# 2. Update Ada's grades to include a new grade of 95
# 3. Remove Grace from the gradebook
# 4. Print each student and their average using .items()
# 5. Dictionary comprehension — student names and their averages:
#    {"Ada": 88.75, "Alan": 68.75, ...}

#1. 
gradebook["Von"] = [40, 45, 38, 50]

#2. 
gradebook["Ada"] = [92, 88, 85, 90, 95]

#3. 
gradebook.pop("Grace")

#4. 
for key,value in gradebook.items():
    print(f"{key}: Average grade = {round(sum(value)/len(value),2)}")

#5. 
average = {key:round(sum(value)/len(value),2) for key,value in gradebook.items()}
print(average)
```
